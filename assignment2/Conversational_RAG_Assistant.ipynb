```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("knowledge.txt")
documents = loader.load()

print(documents)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=200,
    chunk_overlap=20
)

docs = splitter.split_documents(documents)

print(len(docs))
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
from langchain_community.vectorstores import FAISS

vectorstore = FAISS.from_documents(
    docs,
    embeddings
)
```

```python
query = "What are the benefits of RAG?"

results = vectorstore.similarity_search(
    query,
    k=2
)

for doc in results:
    print(doc.page_content)
```

```python
def rag_answer(question):
    
    results = vectorstore.similarity_search(
        question,
        k=2
    )

    context = "\n\n".join(
        [doc.page_content for doc in results]
    )

    prompt = f"""
Answer only from the context below.

Context:
{context}

Question:
{question}
"""

    response = llm.invoke(prompt)

    return response.content
```

```python
print(
    rag_answer(
        "What are the benefits of RAG?"
    )
)
```

```python
print(
    rag_answer(
        "What is LangChain?"
    )
)
```

```python
def create_standalone_question(chat_history, question):

    history_text = "\n".join(
        [msg.content for msg in chat_history[-4:]]
    )

    prompt = f"""
Given the conversation history and the latest question,
rewrite the latest question as a standalone question.

Conversation:
{history_text}

Question:
{question}

Standalone Question:
"""

    response = llm.invoke(prompt)

    return response.content
```

```python
chat_history = [
    HumanMessage(content="What is RAG?")
]
```

```python
standalone = create_standalone_question(
    chat_history,
    "What are its benefits?"
)

print(standalone)
```

```python
def conversational_rag(question, chat_history):

    standalone_question = create_standalone_question(
        chat_history,
        question
    )

    results = vectorstore.similarity_search(
        standalone_question,
        k=2
    )

    context = "\n\n".join(
        [doc.page_content for doc in results]
    )

    prompt = f"""
Answer using only the context.

Context:
{context}

Question:
{standalone_question}
"""

    response = llm.invoke(prompt)

    return response.content
```

```python
chat_history = []

chat_history.append(
    HumanMessage(content="What is RAG?")
)

answer = conversational_rag(
    "What are its benefits?",
    chat_history
)

print(answer)
```

```python
def should_use_tool(question):

    prompt = f"""
You are a router.

If the question requires calculation,
answer ONLY:

TOOL: calculate

Otherwise answer ONLY:

RAG

Question:
{question}
"""

    response = llm.invoke(prompt)

    return response.content.strip()
```

```python
from tools import calculate

def answer_question(question, chat_history):
    route = should_use_tool(question)

    if "calculate" in route.lower():
        expression = question.replace(
            "What is",
            ""
        ).replace("?", "").strip()

        result = calculate(expression)

        return f"The answer is {result}"

    else:
        return conversational_rag(
            question,
            chat_history
        )
```
